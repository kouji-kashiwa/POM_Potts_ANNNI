```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
import numpy as np
import matplotlib.pyplot as plt
import cmath
import random
import tqdm
import os
os.makedirs('data', exist_ok=True)
torch.set_default_dtype(torch.float64)

# Parameters
L = 6
N_site = L**3
dim = 2 * N_site

# 3状態ポッツ模型の定数
omega  = torch.tensor(cmath.exp( 2j * np.pi / 3), dtype=torch.complex128)
omega2 = torch.tensor(cmath.exp(-2j * np.pi / 3), dtype=torch.complex128)

# Parameters
Jc    = 0.1 # coupling constant
kappa = 0.5

tem = 1.0
mu  = 10.0
beta = 1/tem
mass = 10

h_qcd = np.exp(-beta * (mass - mu))      # h_- に対応
h_bar_qcd = np.exp(-beta * (mass + mu))  # h_+ に対応


def complex_logsumexp(w_stack, dim=0):
    """
    w_stack: (3, N_site) のような複素数テンソル
    dim: 和を取る次元
    """
    # 数値的安定性のために実部の最大値を取得
    # .detach() するのは max_real 自体は微分に関与させない（定数シフトとして扱う）ためです
    max_real = torch.max(w_stack.real, dim=dim, keepdim=True)[0].detach()
    
    # max_real を引いてから exp を取る（実部が 0 以下になるので overflow しない）
    # log(sum(exp(w - max))) + max
    res = torch.log(torch.sum(torch.exp(w_stack - max_real), dim=dim)) + max_real.squeeze(dim)
    return res

# 1. 行列作成 (3D) - ANNNP（FFLO的拡張）版
def create_matrix_3d_annnp(L, nnn, mass_param, kappa=0.0):
    N = L**3
    k = torch.zeros(N, N, dtype=torch.complex128)
    # 3次元座標 (x, y, z) から1次元インデックスを取得するヘルパー関数
    def get_idx(x, y, z):
        return (x % L) + L * (y % L) + L * L * (z % L)
    for z in range(L):
        for y in range(L):
            for x in range(L):
                i = get_idx(x, y, z)
                # --- 距離1の隣接サイト（通常通り揃えようとする引力） ---
                x_plus,  x_minus = get_idx(x + 1, y, z), get_idx(x - 1, y, z)
                y_plus,  y_minus = get_idx(x, y + 1, z), get_idx(x, y - 1, z)
                z_plus,  z_minus = get_idx(x, y, z + 1), get_idx(x, y, z - 1)      
                k[i, x_plus] = 1.0; k[i, x_minus] = 1.0
                k[i, y_plus] = 1.0; k[i, y_minus] = 1.0
                k[i, z_plus] = 1.0; k[i, z_minus] = 1.0         
                # --- 距離2の次近接サイト（Z方向のみ、反発力） ---
                if kappa > 0.0:
                    z_plus_2  = get_idx(x, y, z + 2)
                    z_minus_2 = get_idx(x, y, z - 2)
                    k[i, z_plus_2]  = -kappa
                    k[i, z_minus_2] = -kappa
                # --- 自己相互作用（対角成分） ---
                # 理論的な安定性のため、反発力を入れた分だけ対角成分を調整
                k[i, i] = (nnn + 2.0 * kappa) + mass_param          
    return k

# 1. 行列作成 (3D)
def create_matrix_3d_homo(L, nnn, mass_param):
    N = L**3
    k = torch.zeros(N, N, dtype=torch.complex128)
    # 3次元座標 (x, y, z) から1次元インデックスを取得するヘルパー関数
    def get_idx(x, y, z):
        return (x % L) + L * (y % L) + L * L * (z % L)
    for z in range(L):
        for y in range(L):
            for x in range(L):
                i = get_idx(x, y, z)
                # x, y, z 各方向の隣接サイト（周期境界条件）
                x_plus  = get_idx(x + 1, y, z)
                x_minus = get_idx(x - 1, y, z)
                y_plus  = get_idx(x, y + 1, z)
                y_minus = get_idx(x, y - 1, z)
                z_plus  = get_idx(x, y, z + 1)
                z_minus = get_idx(x, y, z - 1)      
                # 隣接サイトとの結合
                k[i, x_plus]  = 1.0
                k[i, x_minus] = 1.0
                k[i, y_plus]  = 1.0
                k[i, y_minus] = 1.0
                k[i, z_plus]  = 1.0
                k[i, z_minus] = 1.0         
                # 自己相互作用（対角成分）
                k[i, i] = nnn + mass_param          
    return k

# for HMC
class hmc:
    def __init__(self, POM, k_mat, Jc, h_qcd, h_bar_qcd, epsilon, ns):
        self.POM = POM
        self.k_mat = k_mat.to(torch.complex128)
        self.Jc = Jc
        self.h_qcd = h_qcd
        self.h_bar_qcd = h_bar_qcd
        self.epsilon = epsilon
        self.ns = ns
    
    def updation(self,x):
        x_old = x.clone()
        p_old = torch.empty_like(x_old).normal_(mean=0.0, std=1.0)
        x_mid, p_mid = self.molecular_dynamics(x_old,p_old)
        x_new, s_new, accept = self.metropolis_test(x_mid,p_mid,x_old,p_old)
        return x_new, s_new, accept
     
    def molecular_dynamics(self,x,p):
        x_new = x.clone()
        p_new = p - 0.5 * self.epsilon * self.daction(x_new)
        for mdstep in range(0, self.ns - 1):
            x_new = x_new + self.epsilon * p_new
            p_new = p_new - self.epsilon * self.daction(x_new)
        x_new = x_new + self.epsilon * p_new
        p_new = p_new - 0.5 * self.epsilon * self.daction(x_new)
        return x_new, p_new
 
    # Metropolis test
    def metropolis_test(self,x_new,p_new,x_old,p_old):
        s_new, energy_new = self.hamiltonian(x_new,p_new)
        s_old, energy_old = self.hamiltonian(x_old,p_old)
        denergy = energy_new - energy_old
        mc_probability = torch.exp(-denergy)
        if torch.rand(1) < torch.min(torch.ones(1),mc_probability):
            accept = 1
            return x_new, s_new, accept
        else:
            accept = 0
            return x_old, s_old, accept   
        
    def hamiltonian(self, x, p):
        with torch.no_grad():
            jacobian_matrix = self.POM.jacobian_matrix_complex(x)
            #jacobian_matrix = self.POM.jacobian_matrix_complex_analytic(x)
            sign, logabsdet = torch.linalg.slogdet(jacobian_matrix)
            phase = torch.angle(sign)
            logdet_J = logabsdet + 1j * phase
            s_eff = self.action(x) - logdet_J
            energy = torch.real(s_eff + 0.5 * torch.sum(p * p))
        return s_eff, energy
    
    def action(self, x):
        x_c = self.POM.complex_function(x)
        u = x_c[:N_site]
        v = x_c[N_site:]
        z_potts = u + 1.0j * v
        z_potts_bar = u - 1.0j * v   
        Kz = torch.mv(self.k_mat, z_potts)
        Kz_bar = torch.mv(self.k_mat, z_potts_bar)
        interaction_term = torch.dot(z_potts_bar, Kz) / (2.0 * self.Jc) 
        W = self.h_qcd + 0.5 * Kz
        W_bar = self.h_bar_qcd + 0.5 * Kz_bar
        w1 = W + W_bar
        w2 = W * omega + W_bar * omega2
        w3 = W * omega2 + W_bar * omega
        max_val = torch.real(w1)
        max_val = torch.where(torch.real(w2) > max_val, torch.real(w2), max_val)
        max_val = torch.where(torch.real(w3) > max_val, torch.real(w3), max_val)        
        trace_term = torch.sum(max_val + torch.log(
            torch.exp(w1 - max_val) + torch.exp(w2 - max_val) + torch.exp(w3 - max_val)
        ))
        return interaction_term - trace_term
    
    def real_action(self, x):
        return torch.real(self.action(x))

    def daction(self, x):
        x_with_grad = x.detach().requires_grad_(True)
        with torch.enable_grad():
            action_val = self.real_action(x_with_grad)
        grad = torch.autograd.grad(action_val, x_with_grad)[0]
        return grad
    
    def scalar_condensate(self, z_complex):
        with torch.no_grad():
            u = z_complex[:N_site]
            v = z_complex[N_site:]
            z_potts = u + 1.0j * v
            z_potts_bar = u - 1.0j * v  # 解析接続された共役変数
            Kz = torch.mv(self.k_mat, z_potts)
            Kz_bar = torch.mv(self.k_mat, z_potts_bar)
            W = self.h_qcd + 0.5 * Kz
            W_bar = self.h_bar_qcd + 0.5 * Kz_bar  # conj() を使わない
            w1 = W + W_bar
            w2 = W * omega + W_bar * omega2
            w3 = W * omega2 + W_bar * omega            
            # Log-Sum-Expと同様の正規化でオーバーフローを防ぐ
            max_val = torch.real(w1)
            max_val = torch.where(torch.real(w2) > max_val, torch.real(w2), max_val)
            max_val = torch.where(torch.real(w3) > max_val, torch.real(w3), max_val)
            e1 = torch.exp(w1 - max_val)
            e2 = torch.exp(w2 - max_val)
            e3 = torch.exp(w3 - max_val)
            # Pの期待値: (1*e1 + ω*e2 + ω^2*e3) / (e1 + e2 + e3)
            return torch.mean((1.0 * e1 + omega * e2 + omega2 * e3) / (e1 + e2 + e3))
        
    def local_scalar_condensate(self, z_complex):
        with torch.no_grad():
            # POM出力からポッツ模型用の補助場を組み立てる
            u = z_complex[:N_site]
            v = z_complex[N_site:]
            z_potts = u + 1.0j * v
            z_potts_bar = u - 1.0j * v 
            Kz = torch.mv(self.k_mat, z_potts)
            Kz_bar = torch.mv(self.k_mat, z_potts_bar)
            W = self.h_qcd + 0.5 * Kz
            W_bar = self.h_bar_qcd + 0.5 * Kz_bar
            
            w1 = W + W_bar
            w2 = W * omega + W_bar * omega2
            w3 = W * omega2 + W_bar * omega            
            
            # Log-Sum-Expと同様の正規化でオーバーフローを防ぐ
            max_val = torch.real(w1)
            max_val = torch.where(torch.real(w2) > max_val, torch.real(w2), max_val)
            max_val = torch.where(torch.real(w3) > max_val, torch.real(w3), max_val)
            
            e1 = torch.exp(w1 - max_val)
            e2 = torch.exp(w2 - max_val)
            e3 = torch.exp(w3 - max_val)
            
            # 平均を取らずに、各サイトのポリヤコフループの期待値 (N_site,) をそのまま返す
            local_P = (1.0 * e1 + omega * e2 + omega2 * e3) / (e1 + e2 + e3)
            return local_P

    def local_polyakov_loops(self, z_complex):
        with torch.no_grad():
            u = z_complex[:N_site]
            v = z_complex[N_site:]
            z_potts = u + 1.0j * v
            z_potts_bar = u - 1.0j * v
            Kz = torch.mv(self.k_mat,z_potts)
            Kz_bar = torch.mv(self.k_mat,z_potts_bar)
            W = (self.h_qcd+ 0.5 * Kz)
            W_bar = (self.h_bar_qcd+ 0.5 * Kz_bar)
            w1 = W + W_bar
            w2 = (omega * W + omega2 * W_bar)
            w3 = (omega2 * W + omega * W_bar)
            # サイトごとに実部の最大値を引く
            max_val = torch.maximum( torch.real(w1), torch.maximum( torch.real(w2), torch.real(w3) ) )
            e1 = torch.exp(w1 - max_val)
            e2 = torch.exp(w2 - max_val)
            e3 = torch.exp(w3 - max_val)
            denominator = e1 + e2 + e3
            local_phi = ( e1 + omega * e2 + omega2 * e3 ) / denominator
            local_phi_bar = ( e1 + omega2 * e2 + omega * e3) / denominator
            return local_phi, local_phi_bar

    def energy(self, x):
        with torch.no_grad():
            return self.action(x)



class Mode(nn.Module):
    def __init__(self,dim,mid):
        super().__init__()
        self.linear1 = nn.Linear(dim, mid)
        self.linear2 = nn.Linear(mid, dim)
        self.mish = nn.Mish()
        # ゼロに近い値で初期化し、初期状態を元の実積分路に近づける
        torch.nn.init.xavier_normal_(self.linear1.weight, gain=0.01)
        torch.nn.init.xavier_normal_(self.linear2.weight, gain=0.01)
        torch.nn.init.zeros_(self.linear1.bias)
        torch.nn.init.zeros_(self.linear2.bias)
    
    def forward(self,x):
        x = self.linear1(x)
        x = self.mish(x)
        x = self.linear2(x)        
        return x

class Model(nn.Module):
    def __init__(self,dim,mid):
        super().__init__()
        self.linear1 = nn.Linear(dim, mid)
        self.linear2 = nn.Linear(mid, dim)
        self.tanh = nn.Tanh()
        # ゼロに近い値で初期化し、初期状態を元の実積分路に近づける
        torch.nn.init.xavier_normal_(self.linear1.weight, gain=0.01)
        torch.nn.init.xavier_normal_(self.linear2.weight, gain=0.01)
        torch.nn.init.zeros_(self.linear1.bias)
        torch.nn.init.zeros_(self.linear2.bias)
    
    def forward(self,x):
        x = self.linear1(x)
        x = self.tanh(x)
        x = self.linear2(x)        
        return x

class CNNModel(nn.Module):
    def __init__(self, L, hidden_channels=8):
        super().__init__()
        self.L = L
        self.conv1 = nn.Conv3d(
            in_channels=2,
            out_channels=hidden_channels,
            kernel_size=3,
            padding=1,
            padding_mode="circular"
        )
        self.conv2 = nn.Conv3d(
            in_channels=hidden_channels,
            out_channels=2,
            kernel_size=3,
            padding=1,
            padding_mode="circular"
        )
        self.activation = nn.Tanh()
        nn.init.xavier_normal_(
            self.conv1.weight,
            gain=0.01
        )
        nn.init.xavier_normal_(
            self.conv2.weight,
            gain=0.01
        )
        nn.init.zeros_(self.conv1.bias)
        nn.init.zeros_(self.conv2.bias)
        
    def forward(self, x):
        # x: (2 * L**3,)
        x_grid = x.reshape(1,2,self.L,self.L,self.L)
        y = self.conv1(x_grid)
        y = self.activation(y)
        y = self.conv2(y)
        return y.reshape(-1)

# Path Optimization Method
class path_optimization(nn.Module): 
    def __init__(self,dim,mid,L,hidden_channels=8):
        super(path_optimization,self).__init__()
        #self.nn_model = Model(dim,mid)
        self.nn_model = CNNModel(L=L,hidden_channels=hidden_channels)

    def forward(self,x):
        return self.nn_model(x)

    # Jacobian calculation (complex value) 
    def jacobian_matrix_complex(self,x,create_graph=False):
        jacobian_matrix = torch.autograd.functional.jacobian(self.transform_complex_to_real,x,create_graph=create_graph)
        jacobian_matrix_reshape = jacobian_matrix.permute(0,2,1).contiguous()
        jacobian_matrix_reshape = jacobian_matrix_reshape.to(torch.float64)
        ans = torch.view_as_complex(jacobian_matrix_reshape).reshape((x.numel(),x.numel()))
        return ans

    def jacobian_matrix_complex_analytic2(self, x):
        # ネットワークのパラメータを取得 (2層モデルを想定)
        W1 = self.nn_model.linear1.weight # shape: (mid, dim)
        b1 = self.nn_model.linear1.bias   # shape: (mid,)
        W2 = self.nn_model.linear2.weight # shape: (dim, mid) 
        # 1. 順伝播の途中計算
        z = torch.matmul(W1, x) + b1
        # 2. Mishの微分 (解析的な計算)
        sp = torch.nn.functional.softplus(z)
        t = torch.tanh(sp)
        s = torch.sigmoid(z)
        dh = t + z * s * (1.0 - t**2) # shape: (mid,)        
        # 3. ニューラルネットワーク部分のヤコビアン (W2 * diag(dh) * W1)
        # ブロードキャストを利用して効率的に計算
        J_nn = torch.matmul(W2 * dh, W1) 
        # 4. 複素化部分を含む全体のヤコビアン: J = I + i * (ig / (ir + 1)) * J_nn
        eye = torch.eye(x.size(0), dtype=torch.complex128, device=x.device)
        coef = self.ig / (self.ir + 1)
        J = eye + 1.0j * coef * J_nn
        return J
    
    def jacobian_matrix_complex_analytic(self, x):
        # ネットワークのパラメータを取得
        W1 = self.nn_model.linear1.weight # shape: (mid, dim)
        b1 = self.nn_model.linear1.bias   # shape: (mid,)
        W2 = self.nn_model.linear2.weight # shape: (dim, mid) 
        # 1. 順伝播の途中計算
        z = torch.matmul(W1, x) + b1
        h = torch.tanh(z)
        # 2. Tanhの微分
        dh = 1.0 - h**2 # shape: (mid,)
        # 3. ニューラルネットワーク部分のヤコビアン (W2 * diag(dh) * W1)
        # dh を W2 の各列に掛けてから W1 と内積をとる
        J_nn = torch.matmul(W2 * dh, W1) 
        # 4. 複素化部分を含む全体のヤコビアン: J = I + i * (ig / (ir + 1)) * J_nn
        eye = torch.eye(x.size(0), dtype=torch.complex128, device=x.device)
        coef = self.ig / (self.ir + 1)
        J = eye + 1.0j * coef * J_nn
        return J
            
    def transform_complex_to_real(self,x):
        z = self.complex_function(x) 
        z = z.unsqueeze(-1)
        return torch.cat(([torch.real(z),torch.imag(z)]),dim=-1)
    
    def complex_function(self,x):
        imag_part = (self.ig/(self.ir+1)) * self.nn_model(x)
        z = torch.complex(
            x,
            imag_part
            ).to(torch.complex128)
        return z
        
    def set_igr(self,ig,ir):
        self.ig = ig; self.ir = ir

class jackknife:
    def __init__(self):
        pass
    
    # Jackknife method without reweighting
    def jackknife_obs(self, z, bin_size): # without reweighting
        ntot = z.numel(); nbin = int(ntot / bin_size)
        ntot = int(nbin*bin_size)
        x = z.detach().cpu().numpy()
        sum_all = np.sum( x[:ntot] , axis=0 )
        bin_ans = [ (sum_all - np.sum( x[c*bin_size:(c+1)*bin_size] , axis=0 )) / ( ntot-bin_size ) for c in range(nbin) ]
        ave = np.mean( bin_ans , axis=0 )
        err = np.mean( np.square(np.real(bin_ans-ave)) + 1.0j*np.square(np.imag(bin_ans-ave)) , axis=0 )
        err = ( np.sqrt( np.real(err) ) + 1.0j*np.sqrt( np.imag(err) ) ) * np.sqrt(nbin-1.0)
        return ave, err

    # Jackknife method with reweighting
    def jackknife_rew(self, z1, z2, bin_size): # with reweighting
        with torch.no_grad():
            ntot = z1.numel(); nbin = int(ntot / bin_size)
            ntot = int(nbin*bin_size)
            x1 = (z1*z2).detach().cpu().numpy()
            x2 = z2.detach().cpu().numpy()
            sum_all1 = np.sum( x1[:ntot] , axis=0 )
            sum_all2 = np.sum( x2[:ntot] , axis=0 )
            bin_ans = [ ((sum_all1 - np.sum( x1[c*bin_size:(c+1)*bin_size] , axis=0 )) / (sum_all2 - np.sum( x2[c*bin_size:(c+1)*bin_size] , axis=0 ))) for c in range(nbin) ]
            ave = np.mean( bin_ans, axis=0 )
            err = np.mean( np.square(np.real(bin_ans-ave)) + 1.0j*np.square(np.imag(bin_ans-ave)) , axis=0 )
            err = ( np.sqrt( np.real(err) ) + 1.0j*np.sqrt( np.imag(err) ) ) * np.sqrt(nbin-1.0)
        return ave, err

    def jackknife_connected_rew(self,corr_history,phi_conf,phi_bar_conf,phase,bin_size):
        with torch.no_grad():
            # ----------------------------------------------------
            # Convert to NumPy
            # ----------------------------------------------------
            corr = (corr_history.detach().cpu().numpy())
            phi = (phi_conf.detach().cpu().numpy())
            phi_bar = (phi_bar_conf.detach().cpu().numpy())
            w = (phase.detach().cpu().numpy())
            # ----------------------------------------------------
            # Number of usable configurations
            # ----------------------------------------------------
            n_conf = corr.shape[0]
            L_z = corr.shape[1]
            n_bin = n_conf // bin_size
            if n_bin < 2:
                raise ValueError("At least two jackknife bins are required.")
            n_used = n_bin * bin_size
            corr = corr[:n_used]
            phi = phi[:n_used]
            phi_bar = phi_bar[:n_used]
            w = w[:n_used]
            # ----------------------------------------------------
            # Weighted total sums
            # ----------------------------------------------------
            sum_w = np.sum(w)
            sum_corr_w = np.sum(corr * w[:, None],axis=0)
            sum_phi_w = np.sum(phi * w)
            sum_phi_bar_w = np.sum(phi_bar * w)
            # ----------------------------------------------------
            # Leave-one-bin-out connected correlators
            # ----------------------------------------------------
            corr_jk_samples = np.empty((n_bin, L_z),dtype=np.complex128)
            for ibin in range(n_bin):
                i_start = ibin * bin_size
                i_end = (ibin + 1) * bin_size
                w_bin = w[i_start:i_end]
                # Common denominator for this jackknife sample
                denominator = (sum_w - np.sum(w_bin))
                if np.abs(denominator) < 1.0e-14:
                    raise ZeroDivisionError("The reweighting denominator is nearly zero "f"for jackknife bin {ibin}.")
                # Full two-point correlator in this JK sample
                corr_bin_sum = np.sum(corr[i_start:i_end] * w_bin[:, None],axis=0)
                corr_jk = (sum_corr_w - corr_bin_sum) / denominator
                # <Phi> in the same JK sample
                phi_bin_sum = np.sum(phi[i_start:i_end] * w_bin)
                phi_jk = (sum_phi_w - phi_bin_sum) / denominator
                # <Phi_bar> in the same JK sample
                phi_bar_bin_sum = np.sum(phi_bar[i_start:i_end] * w_bin)
                phi_bar_jk = (sum_phi_bar_w - phi_bar_bin_sum) / denominator
                # Connected correlator in this JK sample
                corr_jk_samples[ibin] = (corr_jk - phi_jk * phi_bar_jk)
            # ----------------------------------------------------
            # Jackknife central value
            # ----------------------------------------------------
            corr_mean_complex = np.mean(corr_jk_samples,axis=0)
            # ----------------------------------------------------
            # Correct jackknife errors
            # ----------------------------------------------------
            diff_real = (corr_jk_samples.real - corr_mean_complex.real[None, :])
            diff_imag = (corr_jk_samples.imag - corr_mean_complex.imag[None, :])
            corr_err = np.sqrt((n_bin - 1) / n_bin * np.sum( diff_real**2,axis=0))
            corr_err_imag = np.sqrt((n_bin - 1) / n_bin * np.sum(diff_imag**2,axis=0))
        return (corr_mean_complex.real,corr_err,corr_mean_complex.imag,corr_err_imag,corr_jk_samples)

# fixing the seeds for random values
def torch_fix_seed(seed=42):
    # Python random
    random.seed(seed)
    # Numpy
    np.random.seed(seed)
    # Pytorch
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.backends.cudnn.deterministic = True
    torch.use_deterministic_algorithms = True

initial_rand = np.random.rand() 
seed = np.random.get_state()[1][0]
seed = 592064287
torch_fix_seed(seed)
print(seed)



##############
# Parameters #
##############
dir_name_out = 'data/'
file_name_out = '_hmc_data.npy'

# HMC #
mid   = N_site 
nt    = N_site*10
num_channel = 8
nrepl = 1 # replica number
nconf_s = 1000 # configuration number
nconf_f = 1000
ns    = 40  # MD step
nskip = 1 # skip
bin_size  = 50 # bin size
epsilon = 1.0/ns # MD step
lam = 1.0

# Parameters in connectivity matrix
nnn   = 6.0  # tuning parameter
mass_param = 1e-5 # tuning parameter

# k_matの初期化
#k_mat = create_matrix_3d(L, nnn, mass_param)
k_mat = create_matrix_3d_annnp(L, nnn, mass_param, kappa)

# Setup
POM = path_optimization(dim=dim, mid=dim, L=L, hidden_channels=num_channel).double()
hmc = hmc(POM, k_mat, Jc=Jc, h_qcd=h_qcd, h_bar_qcd=h_bar_qcd, epsilon=epsilon, ns=ns)
jackknife = jackknife()

# Learning
itraining = 20
#batch_size = 128
batch_size = 16
epoch_num = 3
lr = 1e-3
gamma = 0.99 # rate of exponential moving average
optimizer = torch.optim.AdamW(POM.parameters(), eps=1e-4, lr=lr) # optimizer
#optimizer = torch.optim.SGD(POM.parameters(), lr=lr, momentum = 0.6, nesterov = True) # optimizer

# initial values for dynamical variables
x          = torch.zeros(nrepl, dim, dtype=torch.float64)
x_temp     = torch.randn(dim, dtype=torch.float64)
s_eff      = torch.zeros(nrepl, dtype=torch.complex128)
s_eff_temp = torch.zeros(1, dtype=torch.complex128)
z          = torch.zeros(dim, dtype=torch.complex128)
z_temp     = torch.zeros(dim, dtype=torch.complex128)

########################
# HMC on original path #
########################
output_data_final = []

jt = 0; ig = 0

# thermalization 
for mcstep in tqdm.tqdm(range(nt)):
    for ir in range(nrepl):
        hmc.POM.set_igr(ig,ir); POM.set_igr(ig,ir)
        x[ir,:], s_eff[ir], accept = hmc.updation(x[ir,:])
     
# main part
temp1 = torch.tensor([], dtype=torch.complex128)
temp2 = torch.tensor([], dtype=torch.complex128)
temp3 = torch.tensor([], dtype=torch.complex128)
obs1  = torch.tensor([], dtype=torch.complex128)
obs2  = torch.tensor([], dtype=torch.complex128)
pha   = torch.tensor([], dtype=torch.complex128)
profile_z_list = []
profile_bar_z_list = []

number = str(jt + 1000)
with open(dir_name_out+number+file_name_out, 'wb') as f:
    
    accept_total = 0
    
    file_name_conf = "data/hist_" + number + ".dat"
    f2 = open(file_name_conf, 'a', newline='\n')

    nconf = nconf_f
    for mcstep in tqdm.tqdm(range(nconf)):

        for i in range(nskip):
            for ir in range(nrepl):
                hmc.POM.set_igr(ig,ir); POM.set_igr(ig,ir)
                x[ir,:], s_eff[ir], accept = hmc.updation(x[ir,:])
                accept_total += accept

        ################# 複素化
        ir = 0; hmc.POM.set_igr(ig,ir); POM.set_igr(ig,ir)
        with torch.no_grad():
            z = POM.complex_function(x[ir,:])
        #################
        output_data = torch.view_as_real(torch.cat([s_eff[ir].unsqueeze(0),z.view(-1)])).detach().numpy()    
        np.save(f,output_data)

        with torch.no_grad():
            temp1 = hmc.scalar_condensate(z)
            temp2 = hmc.energy(x[ir,:])
            temp3 = torch.exp(-1.0j*torch.imag(s_eff[0]))
            obs1 = torch.cat([obs1,temp1.unsqueeze(0)], 0)
            obs2 = torch.cat([obs2,temp2.unsqueeze(0)], 0)
            pha = torch.cat([pha,temp3.unsqueeze(0)], 0)
            # ----------------------------------------
            # Z方向の1次元 Polyakov-loop profile
            # ----------------------------------------
            (local_phi, local_phi_bar) = hmc.local_polyakov_loops(z)
            # get_idx = x + L*y + L^2*z なので、view(L,L,L) の添字順は [z,y,x]
            phi_3d = local_phi.reshape( L, L, L )
            phi_bar_3d = local_phi_bar.reshape( L, L, L )
            # x-y 平面平均 shape = (L,)
            phi_z = torch.mean( phi_3d, dim=(1, 2) )
            phi_bar_z = torch.mean( phi_bar_3d, dim=(1, 2))
            # 現在の1配位分だけ追加
            profile_z_list.append(phi_z.clone())
            profile_bar_z_list.append(phi_bar_z.clone())
        f2.write(str(mcstep)+" ")
        f2.write(str((temp1.item()*temp3.item()).real)+" ")
        f2.write(str((temp1.item()*temp3.item()).imag)+" ")
        f2.write(str(temp3.real.item())+" ")
        f2.write(str(temp3.imag.item())+" ")
        cphase = cmath.phase(temp3.item())
        f2.write(str(cphase)+" ")
        #f2.write(str(x[0,:]))
        f2.write('\n')
    f2.close()
        
# リストを shape = (N_conf, L) の tensor に変換
profile_z_history = torch.stack(profile_z_list, dim=0)
profile_bar_z_history = torch.stack(profile_bar_z_list, dim=0)
N_conf, L_z = profile_z_history.shape
print("original profile_z_history shape =", profile_z_history.shape)
print("original profile_bar_z_history shape =",profile_bar_z_history.shape)
assert profile_z_history.shape == (nconf,L)
assert profile_bar_z_history.shape == (nconf,L)
assert pha.numel() == nconf
np.savez_compressed(
    f"data/profile_z_kappa{kappa}_original.npz",
    phi_z=(profile_z_history.detach().cpu().numpy()),
    phi_bar_z=(profile_bar_z_history.detach().cpu().numpy()),
    phase=(pha.detach().cpu().numpy()))
N_conf, L_z = (profile_z_history.shape)

corr_history = torch.zeros((N_conf, L_z), dtype=torch.complex128)
for iconf in range(N_conf):
    phi_z_conf = profile_z_history[iconf]
    phi_bar_z_conf = profile_bar_z_history[iconf]
    for r in range(L_z):
        phi_bar_shifted = torch.roll(phi_bar_z_conf,shifts=-r,dims=0)
        corr_history[iconf, r] = torch.mean(phi_z_conf * phi_bar_shifted)

phi_conf = torch.mean(profile_z_history,dim=1)
phi_bar_conf = torch.mean(profile_bar_z_history,dim=1)
    
# <Phi>
(corr_mean,corr_err,corr_mean_imag,corr_err_imag,corr_jk_samples) = jackknife.jackknife_connected_rew(corr_history=corr_history,phi_conf=phi_conf,phi_bar_conf=phi_bar_conf,phase=pha,bin_size=bin_size)
# error estimation
e_obs1, obs1_err = jackknife.jackknife_rew(obs1, pha, bin_size)
e_obs2, obs2_err = jackknife.jackknife_rew(obs2, pha, bin_size)
e_pha, pha_err   = jackknife.jackknife_obs(pha, bin_size)
print(0, e_obs1.real, obs1_err.real, e_obs1.imag, obs1_err.imag) 
print(0, e_pha.real, pha_err.real, e_pha.imag, pha_err.imag) 
print(0, np.sqrt(e_pha.real*e_pha.real+e_pha.imag*e_pha.imag),'  accept/reject rate = ', accept_total/(nskip*nconf*nrepl))


print("\n--- 2点空間相関関数 C(r) と 構造因子 S(q) の解析 ---")
corr_mean = np.asarray(corr_mean)
corr_err = np.asarray(corr_err)

# 3. 構造因子 S(q) の計算 (C(r) のフーリエ変換)
# q = 2πk / L
k_indices = np.arange(L_z)
q_values = 2.0 * np.pi * k_indices / L_z
n_bin = corr_jk_samples.shape[0]
Sq_jk_samples = np.zeros((n_bin, L_z),dtype=np.float64)
r_values = np.arange(L_z)
for ibin in range(n_bin):
    for k_index in range(L_z):
        q = q_values[k_index]
        Sq_jk_samples[ibin, k_index] = np.sum( corr_jk_samples[ibin].real * np.cos(q * r_values))
Sq_mean = np.mean(Sq_jk_samples,axis=0)
Sq_err = np.sqrt((n_bin - 1) / n_bin * np.sum(( Sq_jk_samples - Sq_mean[None, :] )**2, axis=0))

np.savez_compressed(
    f"data/correlation_Sq_kappa{kappa}_original.npz",
    correlation_history=(corr_history.detach().cpu().numpy()),
    corr_mean=corr_mean,
    corr_err=corr_err,
    corr_mean_imag=corr_mean_imag,
    corr_err_imag=corr_err_imag,
    corr_jk_samples=corr_jk_samples,
    Sq_jk_samples=Sq_jk_samples,
    phi_z=(profile_z_history.detach().cpu().numpy()),
    phi_bar_z=(profile_bar_z_history.detach().cpu().numpy()),
    phase=pha.detach().cpu().numpy(),
    k_indices=k_indices,
    q_values=q_values,
    Sq_mean=Sq_mean,
    Sq_err=Sq_err,
    bin_size=bin_size,
    nconf=N_conf
)

# --- プロット ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# C(r) のプロット
r_coords = np.arange(L_z)
ax1.errorbar(r_coords, corr_mean, yerr=corr_err, fmt='o-', color='blue', capsize=5)
ax1.set_xlabel('Distance r (along Z-axis)')
ax1.set_ylabel('Correlation Function C(r)')
ax1.set_title('Spatial Correlation Function')
ax1.grid(True)

# S(q) のプロット
ax2.errorbar(k_indices,Sq_mean,yerr=Sq_err,fmt='o-',color='red',capsize=5)
ax2.set_xlabel('Wavenumber index k (q = 2πk/L)')
ax2.set_ylabel('Structure Factor S(q)')
ax2.set_title('Structure Factor (Fourier Transform)')
ax2.set_xticks(k_indices)
ax2.grid(True)

plt.tight_layout()
# kappa の値を変数として持っている前提で保存
plt.savefig(f'data/correlation_Sq_kappa{kappa}_original.png', dpi=300) 
plt.show()




##################################################################################################################
nconf = nconf_s; icount = 0; f3 = open('data/learning.dat', 'a', newline='\n')

save_jt = 0
save_APF = -np.inf

best_corr_history = None
best_corr_mean = None
best_corr_err = None
best_profile_z_history = None
best_profile_bar_z_history = None
best_pha = None
best_model_state = None
best_corr_mean_imag = None
best_corr_err_imag = None
best_corr_jk_samples = None

scheduler = torch.optim.lr_scheduler.ExponentialLR(optimizer,gamma=0.9)
for jt in range(itraining): 
    lam = 1.0
    ig = 1; ir = 0; hmc.POM.set_igr(ig,ir); POM.set_igr(ig,ir)

    # setting for training
    dataset = torch.empty((0,x_temp.numel()+1), dtype=torch.complex128)
    with open(dir_name_out+number+file_name_out, 'rb') as f:
#
        for i in range(nconf):
            output_data2 = np.load(f)
            temp_read = torch.view_as_complex(torch.from_numpy(output_data2))
            dataset = torch.cat((dataset, temp_read.unsqueeze(0)), 0)
    # training    
    apf, apf_abs = torch.tensor(1e-5), torch.tensor(1e-5)

    #POM.train()
    for i in tqdm.tqdm(range(epoch_num)):
        train_data = DataLoader(dataset, batch_size=batch_size, shuffle=True)

        for batch in iter(train_data):
            loss_function = 0
            with torch.no_grad():
                apf_pha0 = apf/apf_abs
                apf_pha0 = apf_pha0/torch.abs(apf_pha0+ 1e-10)
        
            for data in batch:
                x_temp = torch.real(data[1:]); s_eff_old = data[0]
                z_temp = POM.complex_function(x_temp).imag
                
                # 作用の計算
                s = hmc.action(x_temp)
                jacobian_matrix = POM.jacobian_matrix_complex(x_temp,create_graph=True)
                #jacobian_matrix = POM.jacobian_matrix_complex_analytic(x_temp)
                
                # 自動微分を生かすため、no_gradは外す（または必要な部分だけ切り分ける）
                # det_J の代わりに slogdet を使用して数値的安定性を確保
                sign, logabsdet = torch.linalg.slogdet(jacobian_matrix)
                phase = torch.angle(sign)
                logdet_J = logabsdet + 1.0j * phase
                s_eff2 = s - logdet_J
                
                with torch.no_grad():
                    # --- 指数関数の発散を防ぐためのクランプ処理 ---
                    diff_real = torch.real(s_eff2 - s_eff_old)
                    diff_real_clamped = torch.clamp(diff_real, min=-5.0, max=10.0)
                    # 実部だけをクランプした新しい差分を作る
                    s_diff = diff_real_clamped + 1.0j * torch.imag(s_eff2 - s_eff_old)
                    apf     = apf + torch.exp(-s_diff)
                    apf_abs = apf_abs + torch.exp(-diff_real_clamped)
                    apf_pha = torch.exp(-1.0j * torch.imag(s_eff2))
                    rew_fac = torch.exp(-diff_real_clamped)
                    
                loss_function = loss_function + rew_fac * torch.real((1 - apf_pha/apf_pha0) * (-s_eff2)) + lam * torch.dot(z_temp, z_temp)/dim
                
                if loss_function != loss_function:
                    print('Error (NAN)', jt, i, loss_function)
                assert not torch.isnan(loss_function)
                
            optimizer.zero_grad()
            loss_function = loss_function / len(batch)
            loss_function.backward()
            
            torch.nn.utils.clip_grad_norm_(POM.parameters(), max_norm=0.1)
            optimizer.step()
            f3.write(str(icount+1) + " " +str(loss_function.item()))
            f3.write('\n')
            icount += 1
            
            apf, apf_abs = gamma*apf, gamma*apf_abs

    scheduler.step()
    
    number = str(jt + 1 + 1000)
    torch.save(POM.state_dict(), 'data/model_weight'+number+'.pth')
            
    ##########################################################
    # HMC on modified path with probability of modified path #
    ##########################################################
    # thermalization
    #POM.eval()
    accept_replica = 0
    for mcstep in tqdm.tqdm(range(nt)):
        for ir in range(nrepl):
            hmc.POM.set_igr(ig,ir); POM.set_igr(ig,ir)
            x[ir,:], s_eff[ir], accept = hmc.updation(x[ir,:])
            
        if nrepl > 1:
            for i in range(nrepl-1):
                ir = (nrepl - 1) - i; hmc.POM.set_igr(ig,ir); POM.set_igr(ig,ir)
                # original weight
                ds_or = torch.real(- s_eff[ir-1] - s_eff[ir])
                # exchange path 1 ( energy and second x[.] in POM.hamitonian are dummies)
                s_eff_i, energy = hmc.hamiltonian(x[ir-1,:],x[ir-1,:])
                # exchange path 2 ( energy and second x[.] in POM.hamitonian are dummies)
                hmc.POM.set_igr(ig,ir-1); POM.set_igr(ig,ir-1)
                s_eff_j, energy = hmc.hamiltonian(x[ir,:],x[ir,:])
                # exchange weight
                ds_ex = torch.real(- s_eff_i - s_eff_j)
                # transfer probability
                probability = torch.exp(ds_ex)/torch.exp(ds_or)
                if torch.rand(1) < torch.min(torch.ones(1),probability):
                    x_temp[:] = x[ir-1,:].clone()
                    x[ir-1,:] = x[ir  ,:].clone()
                    x[ir  ,:] = x_temp[:].clone()
                    s_eff[ir  ], energy = hmc.hamiltonian(x[ir  ],x[ir  ])
                    s_eff[ir-1], energy = hmc.hamiltonian(x[ir-1],x[ir-1])
       
    # main part
    temp1 = torch.tensor([], dtype=torch.complex128)
    temp2 = torch.tensor([], dtype=torch.complex128)
    temp3 = torch.tensor([], dtype=torch.complex128)
    obs1  = torch.tensor([], dtype=torch.complex128)
    obs2  = torch.tensor([], dtype=torch.complex128)
    pha   = torch.tensor([], dtype=torch.complex128)
    profile_z_list = []
    profile_bar_z_list = []
    
    with open(dir_name_out+number+file_name_out, 'wb') as f:
        accept_total = 0
        
        number = str(jt + 1 + 1000)
        file_name_conf = "data/hist_" + number + ".dat"
        f2 = open(file_name_conf, 'a', newline='\n')            
        
        if jt == itraining - 1:
            nconf = nconf_f
        if jt < itraining - 1:
            nconf = nconf_s
    
        for mcstep in tqdm.tqdm(range(nconf)):
            for i in range(nskip):
                
                # parallel tempering
                for ir in range(nrepl):
                    hmc.POM.set_igr(ig,ir); POM.set_igr(ig,ir)
                    x[ir,:], s_eff[ir], accept = hmc.updation(x[ir,:])
                    accept_total += accept
                    
                if nrepl > 1:
                    for i in range(nrepl-1):
                        ir = (nrepl - 1) - i; hmc.POM.set_igr(ig,ir); POM.set_igr(ig,ir)
                        # original weight
                        ds_or = torch.real(- s_eff[ir-1] - s_eff[ir])
                        # exchange path 1 ( energy and second x[.] in POM.hamitonian are dummies)
                        s_eff_i, energy = hmc.hamiltonian(x[ir-1,:],x[ir-1,:])
                        # exchange path 2 ( energy and second x[.] in POM.hamitonian are dummies)
                        hmc.POM.set_igr(ig,ir-1); POM.set_igr(ig,ir-1)
                        s_eff_j, energy = hmc.hamiltonian(x[ir,:],x[ir,:])
                        # exchange weight
                        ds_ex = torch.real(- s_eff_i - s_eff_j)
                        # transfer probability
                        probability = torch.exp(ds_ex)/torch.exp(ds_or)
                        if torch.rand(1) < torch.min(torch.ones(1),probability):
                            x_temp[:] = x[ir-1,:].clone()
                            x[ir-1,:] = x[ir  ,:].clone()
                            x[ir  ,:] = x_temp[:].clone()
                            s_eff[ir  ], energy = hmc.hamiltonian(x[ir  ],x[ir  ])
                            s_eff[ir-1], energy = hmc.hamiltonian(x[ir-1],x[ir-1])
                            accept_replica += 1
                    
            ################################################################ complexification
            ir = 0; hmc.POM.set_igr(ig,ir); POM.set_igr(ig,ir)
            with torch.no_grad():
                z = POM.complex_function(x[0,:])
            ################################################################
            output_data = torch.view_as_real(torch.cat([s_eff[0].unsqueeze(0),z.view(-1)])).detach().numpy()
            np.save(f,output_data)     
        
            with torch.no_grad():
                temp1 = hmc.scalar_condensate(z)
                temp2 = hmc.energy(x[ir,:])
                temp3 = torch.exp(-1.0j*torch.imag(s_eff[0]))
                obs1 = torch.cat([obs1,temp1.unsqueeze(0)], 0)
                obs2 = torch.cat([obs2,temp2.unsqueeze(0)], 0)
                pha = torch.cat([pha,temp3.unsqueeze(0)], 0)
                # ----------------------------------------
                # Z方向の1次元 Polyakov-loop profile
                # ----------------------------------------
                (local_phi, local_phi_bar) = hmc.local_polyakov_loops(z)
                # get_idx = x + L*y + L^2*z なので、view(L,L,L) の添字順は [z,y,x]
                phi_3d = local_phi.reshape( L, L, L )
                phi_bar_3d = local_phi_bar.reshape( L, L, L )
                # x-y 平面平均 shape = (L,)
                phi_z = torch.mean( phi_3d, dim=(1, 2) )
                phi_bar_z = torch.mean( phi_bar_3d, dim=(1, 2))
                # 現在の1配位分だけ追加
                profile_z_list.append(phi_z.clone())
                profile_bar_z_list.append(phi_bar_z.clone())
            
            f2.write(str(mcstep)+" ")
            f2.write(str((temp1.item()*temp3.item()).real)+" ")
            f2.write(str((temp1.item()*temp3.item()).imag)+" ")
            f2.write(str((temp2.item()*temp3.item()).real)+" ")
            f2.write(str((temp2.item()*temp3.item()).imag)+" ")
            f2.write(str(temp3.real.item())+" ")
            f2.write(str(temp3.imag.item())+" ")
            cphase = cmath.phase(temp3.item())
            f2.write(str(cphase) + " ")
            f2.write("\n")
            
        f2.close()

    profile_z_history = torch.stack(profile_z_list,dim=0)
    profile_bar_z_history = torch.stack(profile_bar_z_list,dim=0)
    N_conf, L_z = profile_z_history.shape
    print("iteration",jt + 1,"profile_z_history shape =",profile_z_history.shape)
    print("iteration",jt + 1,"profile_bar_z_history shape =",profile_bar_z_history.shape)
    assert profile_z_history.shape == (nconf,L)
    assert profile_bar_z_history.shape == (nconf,L)
    assert pha.numel() == nconf
    np.savez_compressed(f"data/profile_z_kappa{kappa}_"f"iteration{jt + 1:03d}.npz",phi_z=(profile_z_history.detach().cpu().numpy()),phi_bar_z=(profile_bar_z_history.detach().cpu().numpy()),phase=(pha.detach().cpu().numpy()))

# ============================================================
# Correlation function at the current training iteration
# ============================================================
    corr_history = torch.zeros((N_conf, L_z),dtype=torch.complex128)
    for iconf in range(N_conf):
        phi_z_conf = profile_z_history[iconf]
        phi_bar_z_conf = profile_bar_z_history[iconf]
        for r in range(L_z):
            phi_bar_shifted = torch.roll(phi_bar_z_conf,shifts=-r,dims=0)
            corr_history[iconf, r] = torch.mean(phi_z_conf * phi_bar_shifted)
    # Reweighted one-point functions
    phi_conf = torch.mean(profile_z_history,dim=1)
    phi_bar_conf = torch.mean(profile_bar_z_history,dim=1)

    (corr_mean,corr_err,corr_mean_imag,corr_err_imag,corr_jk_samples) = jackknife.jackknife_connected_rew(corr_history=corr_history,phi_conf=phi_conf,phi_bar_conf=phi_bar_conf,phase=pha,bin_size=bin_size)

    # ============================================================
    # Observables and average phase factor at the current iteration
    # ============================================================
    e_obs1, obs1_err = jackknife.jackknife_rew(obs1,pha,bin_size)
    e_obs2, obs2_err = jackknife.jackknife_rew(obs2,pha,bin_size)
    e_pha, pha_err = jackknife.jackknife_obs(pha,bin_size)
    apf_value = float(np.abs(e_pha))
    print(jt+1, e_obs1.real, obs1_err.real, e_obs1.imag, obs1_err.imag) 
    print(jt+1, e_pha.real, pha_err.real, e_pha.imag, pha_err.imag) 
    print(jt+1, np.sqrt(e_pha.real*e_pha.real+e_pha.imag*e_pha.imag),'  accept/reject rate = ', accept_total/(nskip*nconf*nrepl))

    # ========================================================
    # Save the current data if this is the best path
    # ========================================================
    if apf_value > save_APF:
        save_jt = jt + 1
        save_APF = apf_value
        best_corr_history = (corr_history.detach().cpu().clone())
        best_corr_mean = corr_mean.copy()
        best_corr_err = corr_err.copy()
        best_corr_mean_imag = (corr_mean_imag.copy())
        best_corr_err_imag = (corr_err_imag.copy())
        best_corr_jk_samples = (corr_jk_samples.copy())
        best_profile_z_history = (profile_z_history.detach().cpu().clone())
        best_profile_bar_z_history = (profile_bar_z_history.detach().cpu().clone())
        best_pha = (pha.detach().cpu().clone())
        best_model_state = {key: value.detach().cpu().clone() for key, value in POM.state_dict().items()}
        torch.save({"iteration": save_jt,"APF": save_APF,"model_state_dict": best_model_state},f"data/model_weight_best_"f"kappa{kappa}.pth")
        print(
            "Best path updated:",
            "iteration =",
            save_jt,
            "APF =",
            save_APF
        )
    # Current-iteration output
    print(jt + 1,"APF =",apf_value,"acceptance rate =", accept_total / (nskip * nconf * nrepl))

f3.close()

# ============================================================
# Restore the best-path data after all training iterations
# ============================================================
if best_corr_history is None:
    raise RuntimeError("Best-path data were not saved.")
print("Selected best path:","iteration =",save_jt,"APF =",save_APF)
corr_history = best_corr_history
corr_mean = best_corr_mean
corr_err = best_corr_err
corr_mean_imag = best_corr_mean_imag
corr_err_imag = best_corr_err_imag
corr_jk_samples = best_corr_jk_samples
profile_z_history = best_profile_z_history
profile_bar_z_history = best_profile_bar_z_history
pha = best_pha
N_conf, L_z = profile_z_history.shape

k_indices = np.arange(L_z)
q_values = 2.0 * np.pi * k_indices / L_z
n_bin = corr_jk_samples.shape[0]
Sq_jk_samples = np.zeros((n_bin, L_z),dtype=np.float64)
r_values = np.arange(L_z)
for ibin in range(n_bin):
    for k_index in range(L_z):
        q = q_values[k_index]
        Sq_jk_samples[ibin, k_index] = np.sum( corr_jk_samples[ibin].real * np.cos(q * r_values))
Sq_mean = np.mean(Sq_jk_samples,axis=0)
Sq_err = np.sqrt((n_bin - 1) / n_bin * np.sum(( Sq_jk_samples - Sq_mean[None, :] )**2, axis=0))
    
np.savez_compressed(
    f"data/correlation_Sq_kappa{kappa}_modified_best.npz",
    correlation_history=(corr_history.detach().cpu().numpy()),
    corr_mean=corr_mean,
    corr_err=corr_err,
    corr_mean_imag=corr_mean_imag,
    corr_err_imag=corr_err_imag,
    corr_jk_samples=corr_jk_samples,
    Sq_jk_samples=Sq_jk_samples,
    phi_z=(profile_z_history.detach().cpu().numpy()),
    phi_bar_z=(profile_bar_z_history.detach().cpu().numpy()),
    phase=pha.detach().cpu().numpy(),
    k_indices=k_indices,
    q_values=q_values,
    Sq_mean=Sq_mean,
    Sq_err=Sq_err,
    best_iteration=save_jt,
    best_APF=save_APF,
    bin_size=bin_size,
    nconf=N_conf
)

fig, (ax1, ax2) = plt.subplots(1,2,figsize=(12, 5))
r_coords = np.arange(L_z)
ax1.errorbar(r_coords,corr_mean,yerr=corr_err,fmt="o-",color="blue",capsize=5)
ax1.set_xlabel("Distance r along z axis")
ax1.set_ylabel("Correlation function C(r)")
ax1.set_title("Spatial correlation function")
ax1.set_xticks(r_coords)
ax1.grid(True)
ax2.errorbar(k_indices,Sq_mean,yerr=Sq_err,fmt='o-',color='red',capsize=5)
ax2.set_xlabel("Wavenumber index k")
ax2.set_ylabel("Structure factor S(q)")
ax2.set_title("Structure factor")
ax2.set_xticks(k_indices)
ax2.grid(True)
plt.tight_layout()
plt.savefig(f"data/correlation_Sq_"f"kappa{kappa}_modified.png",dpi=300)
plt.show()
```
